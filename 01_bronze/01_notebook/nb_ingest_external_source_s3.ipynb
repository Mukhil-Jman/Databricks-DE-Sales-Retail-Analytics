```python
meta_data = spark.read.format('csv').option('header', 'true').option('inferschema', 'false').load('s3://amzn-s3-databricks-ingestion/metadata.csv')
display(meta_data)
```

```python
for data in meta_data.collect():
    file_path="s3://amzn-s3-databricks-ingestion/"+str(data.file_name+"."+data.extension)
    table = spark.read.format('csv').option('header', 'true').option('inferschema', 'false').load(file_path)
    table.write.mode("overwrite").option("delta.columnMapping.mode", "name").format("delta").saveAsTable(f"cl_mp_de.01_bronze.{data.table_name}")
display(spark.sql("SHOW TABLES"))
```
