```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("MP-DE").getOrCreate()
```

```python
df = spark.read.option('header', 'true').table('cl_mp_de.`01_bronze`.bronze_product')
display(df)
```

### Standardizing column names

```python
df_standarized = df.withColumnRenamed('Product#ID', 'product_id') \
    .withColumnRenamed('Item Name_Description', 'item_name') \
    .withColumnRenamed('[Category_Type]', 'category_type') \
    .withColumnRenamed('Base_Cost', 'base_cost') \
    .withColumnRenamed('@Marked_Price@', 'marked_price')
display(df_standarized)
```

### Casting marked_price as double

```python
df_clean = df_standarized.withColumn('base_cost', df_standarized['base_cost'].cast('double')) \
    .withColumn('marked_price', df_standarized['marked_price'].cast('double'))
display(df_clean)
```

```python
df_clean.write.mode('overwrite').option('overwriteSchema', 'true').saveAsTable('cl_mp_de.`02_silver`.silver_product')
```
