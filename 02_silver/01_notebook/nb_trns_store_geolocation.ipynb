```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("MP-DE").getOrCreate()
```

```python
df = spark.read.table('cl_mp_de.`01_bronze`.bronze_stores_geolocation_list')
display(df)
```

```python
df_standardized = df.withColumnRenamed('ST_ID!!', 'store_id') \
                    .withColumnRenamed('Location_Name_&_Address', 'location_name_and_address') \
                    .withColumnRenamed('City/Region', 'city_and_region') \
                    .withColumnRenamed('Mgr_Contact#', 'manager_contact')
display(df_standardized)
```

```python
from pyspark.sql import functions as sf
# df_clean = df_standardized.withColumn('city_and_region', df_standardized['city_and_region'].cast('string')) \
df_clean = df_standardized.withColumn('city_and_region', sf.regexp_replace('city_and_region', '_', ' - '))
display(df_clean)
```

```python
df_clean.write.mode("overwrite").option('overwriteSchema', 'true').saveAsTable("cl_mp_de.`02_silver`.silver_stores_geolocation_list")
```
