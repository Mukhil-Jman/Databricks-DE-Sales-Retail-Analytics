```sql
DROP TABLE IF EXISTS cl_mp_de.`03_gold_dim`.dim_date
```

```sql

CREATE TABLE cl_mp_de.`03_gold_dim`.dim_date (
  date_key INT PRIMARY KEY,
  full_date DATE,
  day INT,
  month INT,
  quarter INT,
  year INT
)
```

```sql
SELECT DISTINCT(JOINED_DATE) FROM cl_mp_de.`02_silver`.silver_customer
```

```sql
INSERT INTO cl_mp_de.`03_gold_dim`.dim_date
SELECT DISTINCT
  CAST(date_format(joined_date, 'yyyyMMdd') as INT) AS date_key,
  joined_date as full_date,
  day(joined_date) as day,
  month(joined_date) as month,
  quarter(joined_date) as quarter,
  year(joined_date) as year
FROM cl_mp_de.`02_silver`.silver_customer
```

```sql
INSERT INTO cl_mp_de.`03_gold_dim`.dim_date
SELECT DISTINCT
  CAST(date_format(last_audit_date, 'yyyyMMdd') as INT) AS date_key,
  last_audit_date as full_date,
  day(last_audit_date) as day,
  month(last_audit_date) as month,
  quarter(last_audit_date) as quarter,
  year(last_audit_date) as year
FROM cl_mp_de.`02_silver`.silver_inventory
```
