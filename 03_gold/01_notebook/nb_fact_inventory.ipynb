```sql
CREATE TABLE cl_mp_de.`03_gold_fact`.fact_inventory (
  inventory_key BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  date_key INT,
  product_key BIGINT,
  store_key BIGINT,
  stock_on_hand INT  
)
```

```sql
INSERT INTO cl_mp_de.`03_gold_fact`.fact_inventory (
  date_key,
  product_key,
  store_key,
  stock_on_hand
)
SELECT
  CAST(date_format(i.last_audit_date, 'yyyyMMdd') AS INT) as date_key,
  dp.product_key,
  ds.store_key,
  i.stock_on_hand
FROM
  cl_mp_de.`02_silver`.silver_inventment_levels i
LEFT JOIN cl_mp_de.`03_gold_dim`.dim_product dp
ON
  i.sku_id = dp.product_id
LEFT JOIN cl_mp_de.`03_gold_dim`.dim_store ds
ON  ds.store_id = i.store_id
```
