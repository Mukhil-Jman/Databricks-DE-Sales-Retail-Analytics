```sql

CREATE TABLE cl_mp_de.`03_gold_fact`.fact_sales (
  sales_key BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  date_key INT,
  customer_key BIGINT,
  product_key BIGINT,
  store_key BIGINT,
  transaction_id VARCHAR(255),
  quantity_sold INT,
  unit_price DECIMAL(10,2),
  total_amount DECIMAL(10,2)
)
```

```sql
DROP TABLE cl_mp_de.`03_gold_fact`.fact_sales;

```

```sql
TRUNCATE TABLE cl_mp_de.`03_gold_fact`.fact_sales;
```

```sql
INSERT INTO cl_mp_de.`03_gold_fact`.fact_sales (
  date_key,
  customer_key,
  product_key,
  store_key,
  transaction_id,
  quantity_sold,
  unit_price,
  total_amount
)
SELECT
  CAST(date_format(s.date, 'yyyyMMdd') as INT) as date_key,
  dc.customer_key,
  dp.product_key,
  ds.store_key,
  s.transaction_id,
  s.quantity_sold,
  s.unit_price,
  (s.quantity_sold * s.unit_price) as total_amount
FROM cl_mp_de.`02_silver`.silver_sales_details s
LEFT JOIN cl_mp_de.`03_gold_dim`.dim_customer dc
  ON s.customer_id = dc.customer_id
LEFT JOIN cl_mp_de.`03_gold_dim`.dim_product dp
  ON s.product_id = dp.product_id
LEFT JOIN cl_mp_de.`03_gold_dim`.dim_store ds
  ON s.store_id = ds.store_id
```

```sql
SELECT * FROM cl_mp_de.`03_gold_fact`.fact_sales ORDER BY transaction_id;
```
