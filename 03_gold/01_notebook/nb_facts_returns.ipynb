```sql
CREATE TABLE cl_mp_de.`03_gold_fact`.fact_returns (
  return_key BIGINT GENERATED ALWAYS AS IDENTITY PRIMARY KEY,
  date_key INT,
  reason_key BIGINT,
  transaction_id VARCHAR(255),
  return_id VARCHAR(255)
)
```

```sql
INSERT INTO cl_mp_de.`03_gold_fact`.fact_returns (
  date_key,
  reason_key,
  transaction_id,
  return_id
)
SELECT 
  CAST(date_format(r.return_date, 'yyyyMMdd') AS INT) as date_key,
  dr.reason_key,
  r.original_transaction_id as transaction_id,
  r.return_id as return_id
FROM
  cl_mp_de.`02_silver`.silver_return_transaction_logs r
LEFT JOIN cl_mp_de.`03_gold_dim`.dim_reason dr
  ON r.rsn_code = dr.rsn_code
```
